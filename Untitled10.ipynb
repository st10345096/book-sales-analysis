# Clean basic data issues


```python
# Keep the original data unchanged and work on a copy.
books = df.drop(columns=["index"]).copy()

# Fill missing titles and languages with clear placeholder values.
books["Book Name"] = books["Book Name"].fillna("Unknown title")
books["language_code"] = books["language_code"].fillna("Unknown")

# Fill the single missing year with the median publishing year.
median_year = books["Publishing Year"].median()
books["Publishing Year"] = books["Publishing Year"].fillna(median_year)

# Remove accidental spaces from text categories.
for column in ["Author_Rating", "genre", "Publisher ", "language_code"]:
    books[column] = books[column].astype(str).str.strip()

print("Cleaned dataset shape:", books.shape)
print("Remaining missing values:", books.isna().sum().sum())
print("Exact duplicate rows:", books.duplicated().sum())
```

# Define the bestseller proxy
Label books in the top 20% by units sold as Best Seller (proxy). This gives a sales-based target, but the cutoff is dataset-specific, not a universal bestseller standard.

```python
# Set the threshold at the 80th percentile of units sold.
units_sold_cutoff = books["units sold"].quantile(0.80)

# Label books at or above the cutoff as proxy bestsellers.
books["Best_Seller_Proxy"] = (
    books["units sold"] >= units_sold_cutoff
).astype(int)

print(f"Top-20% units-sold cutoff: {units_sold_cutoff:,.0f}")
print("\nProxy class counts (0 = below cutoff, 1 = meets cutoff):")
display(books["Best_Seller_Proxy"].value_counts().sort_index())

print("Positive share:", round(books["Best_Seller_Proxy"].mean(), 3))
```

## Select the predictors

```python
# Remove extra spaces from column names.
books.columns = books.columns.str.strip()

# Use book details as predictors; exclude direct sales outcomes and names.
predictor_columns = [
    "Publishing Year",
    "language_code",
    "Author_Rating",
    "genre",
    "sale price",
    "Publisher"
]

X = books[predictor_columns].copy()
y = books["Best_Seller_Proxy"].copy()

print("Predictors:")
print(predictor_columns)
print("\nTarget class counts:")
print(y.value_counts().sort_index())
```

Target: 856 negative, 214 positive. Split into train, validation, and test sets with stratification to preserve class balance. Fit the encoder and scaler on training data only.

# Split and encode the data

```python
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

numeric_columns = ["Publishing Year", "sale price"]
categorical_columns = [
    "language_code",
    "Author_Rating",
    "genre",
    "Publisher"
]

# Hold out test data for the final evaluation.
X_train_validation, X_test_raw, y_train_validation, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# Create validation data for model and cutoff selection.
X_train_raw, X_validation_raw, y_train, y_validation = train_test_split(
    X_train_validation,
    y_train_validation,
    test_size=0.20,
    random_state=42,
    stratify=y_train_validation
)

# Scale numeric fields and one-hot encode categories.
preprocessor = ColumnTransformer(
    transformers=[
        ("numeric", StandardScaler(), numeric_columns),
        ("categorical", OneHotEncoder(handle_unknown="ignore"), categorical_columns)
    ]
)

# Fit preprocessing on training data only, then transform the other sets.
X_train = preprocessor.fit_transform(X_train_raw)
X_validation = preprocessor.transform(X_validation_raw)
X_test = preprocessor.transform(X_test_raw)

# Convert labels to NumPy arrays for the from-scratch model.
y_train = y_train.to_numpy()
y_validation = y_validation.to_numpy()
y_test = y_test.to_numpy()

print("Training shape:", X_train.shape)
print("Validation shape:", X_validation.shape)
print("Test shape:", X_test.shape)
print("Positive share — train:", round(y_train.mean(), 3))
print("Positive share — validation:", round(y_validation.mean(), 3))
print("Positive share — test:", round(y_test.mean(), 3))
```

The split is balanced across all three sets. Next, train logistic regression from scratch with higher weights on positive examples, since they are only 20% of the data.

# Train logistic regression

```python
import numpy as np

# Convert model scores into probabilities.
def sigmoid(scores):
    scores = np.clip(scores, -35, 35)
    return 1 / (1 + np.exp(-scores))

# Fit logistic regression using weighted loss and gradient descent.
def train_weighted_logistic_regression(
    X, y, learning_rate=0.1, epochs=1000, l2=0.01
):
    y = np.asarray(y, dtype=float)
    number_of_rows, number_of_features = X.shape

    # Give positive examples more weight.
    positive_weight = (len(y) - y.sum()) / y.sum()
    sample_weights = np.where(y == 1, positive_weight, 1.0)

    # Start with zero weights and intercept.
    weights = np.zeros(number_of_features)
    bias = 0.0

    # Record starting loss.
    initial_probabilities = np.full(number_of_rows, 0.5)
    initial_loss = -np.mean(
        sample_weights * (
            y * np.log(initial_probabilities) +
            (1 - y) * np.log(1 - initial_probabilities)
        )
    )

    # Update weights and intercept to reduce weighted loss.
    for _ in range(epochs):
        probabilities = sigmoid(X @ weights + bias)
        errors = sample_weights * (probabilities - y)

        gradient_weights = np.asarray(X.T @ errors).ravel() / number_of_rows
        gradient_weights += l2 * weights
        gradient_bias = errors.mean()

        weights -= learning_rate * gradient_weights
        bias -= learning_rate * gradient_bias

    # Record final loss.
    probabilities = sigmoid(X @ weights + bias)
    final_loss = -np.mean(
        sample_weights * (
            y * np.log(np.clip(probabilities, 1e-15, 1)) +
            (1 - y) * np.log(np.clip(1 - probabilities, 1e-15, 1))
        )
    )

    return weights, bias, positive_weight, initial_loss, final_loss

weights_model, bias_model, positive_weight, initial_loss, final_loss = (
    train_weighted_logistic_regression(X_train, y_train)
)

print(f"Positive-class weight: {positive_weight:.2f}")
print(f"Initial weighted loss: {initial_loss:.4f}")
print(f"Final weighted loss: {final_loss:.4f}")
```

Weighted loss fell from 1.1086 to 1.0568. Choose the probability cutoff on the validation set by maximum F1. The validation set is small, so the cutoff may be sensitive to individual books.

## Choose the prediction cutoff

```python
from sklearn.metrics import precision_recall_curve

# Predict validation probabilities.
validation_probabilities = sigmoid(X_validation @ weights_model + bias_model)

# Calculate precision and recall across possible cutoffs.
precision_values, recall_values, cutoff_values = precision_recall_curve(
    y_validation, validation_probabilities
)

# Calculate F1 for each available cutoff.
f1_values = (
    2 * precision_values[1:] * recall_values[1:] /
    (precision_values[1:] + recall_values[1:] + 1e-12)
)

# Select the cutoff with the highest validation F1.
best_index = np.argmax(f1_values)
model_cutoff = cutoff_values[best_index]

print(f"Selected cutoff: {model_cutoff:.3f}")
print(f"Validation precision: {precision_values[best_index + 1]:.3f}")
print(f"Validation recall: {recall_values[best_index + 1]:.3f}")
print(f"Validation F1: {f1_values[best_index]:.3f}")
```

Selected cutoff: 0.530. Validation precision is 0.321 and recall 0.500. Apply this cutoff once to the held-out test set.

## Evaluate on the test set

```python
from sklearn.metrics import (
    accuracy_score,
    balanced_accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score,
    confusion_matrix
)

# Predict test probabilities and assign proxy classes.
test_probabilities = sigmoid(X_test @ weights_model + bias_model)
test_predictions = (test_probabilities >= model_cutoff).astype(int)

# Report test performance.
print(f"Cutoff: {model_cutoff:.3f}")
print(f"Accuracy: {accuracy_score(y_test, test_predictions):.3f}")
print(f"Balanced accuracy: {balanced_accuracy_score(y_test, test_predictions):.3f}")
print(f"Precision: {precision_score(y_test, test_predictions, zero_division=0):.3f}")
print(f"Recall: {recall_score(y_test, test_predictions, zero_division=0):.3f}")
print(f"F1 score: {f1_score(y_test, test_predictions, zero_division=0):.3f}")
print(f"ROC-AUC: {roc_auc_score(y_test, test_probabilities):.3f}")
print(f"PR-AUC: {average_precision_score(y_test, test_probabilities):.3f}")

print("\nConfusion matrix (rows = true class, columns = predicted class):")
print(confusion_matrix(y_test, test_predictions))
```

Test results: found 22 of 43 positives (recall ≈ 0.51), correctly rejected 115 of 171 negatives, with 56 false positives. PR-AUC is 0.266 vs. a 0.201 baseline; ROC-AUC is 0.571, near random. The model beats baseline slightly but is not a reliable predictor, and the test set is small (43 positives).

#### Group rare publisher labels

```python
# Keep publisher labels with at least 20 training records.
publisher_counts = X_train_raw["Publisher"].value_counts()
common_publishers = publisher_counts[publisher_counts >= 20].index

# Map less common or unseen publishers to "Other".
def group_rare_publishers(frame):
    grouped = frame.copy()
    grouped["Publisher"] = grouped["Publisher"].where(
        grouped["Publisher"].isin(common_publishers),
        "Other"
    )
    return grouped

X_train_grouped = group_rare_publishers(X_train_raw)
X_validation_grouped = group_rare_publishers(X_validation_raw)
X_test_grouped = group_rare_publishers(X_test_raw)

print("Publisher labels grouped as Other:")
print(publisher_counts[publisher_counts < 20])

print("\nPublisher categories after grouping:")
print(sorted(X_train_grouped["Publisher"].unique()))
```

The two rare publisher labels were based on only five training books combined, so grouping them as Other is reasonable. Retrain with the same settings and compare against the earlier model’s test results as the baseline.

#### Retrain with grouped publishers

```python
# Refit preprocessing with the grouped publisher categories.
preprocessor_grouped = ColumnTransformer(
    transformers=[
        ("numeric", StandardScaler(), numeric_columns),
        ("categorical", OneHotEncoder(handle_unknown="ignore"), categorical_columns)
    ]
)

X_train_grouped_encoded = preprocessor_grouped.fit_transform(X_train_grouped)
X_validation_grouped_encoded = preprocessor_grouped.transform(X_validation_grouped)
X_test_grouped_encoded = preprocessor_grouped.transform(X_test_grouped)

# Train using the same logistic regression settings.
(
    weights_grouped,
    bias_grouped,
    positive_weight_grouped,
    initial_loss_grouped,
    final_loss_grouped
) = train_weighted_logistic_regression(
    X_train_grouped_encoded,
    y_train,
    learning_rate=0.1,
    epochs=1000,
    l2=0.01
)

print("Training features:", X_train_grouped_encoded.shape[1])
print(f"Positive-class weight: {positive_weight_grouped:.2f}")
print(f"Initial weighted loss: {initial_loss_grouped:.4f}")
print(f"Final weighted loss: {final_loss_grouped:.4f}")
```

```python
# Predict validation probabilities from the grouped-publisher model.
validation_probabilities_grouped = sigmoid(
    X_validation_grouped_encoded @ weights_grouped + bias_grouped
)

# Find the cutoff with the highest validation F1 score.
precision_grouped, recall_grouped, cutoffs_grouped = precision_recall_curve(
    y_validation, validation_probabilities_grouped
)

f1_grouped = (
    2 * precision_grouped[1:] * recall_grouped[1:] /
    (precision_grouped[1:] + recall_grouped[1:] + 1e-12)
)

best_index_grouped = np.argmax(f1_grouped)
cutoff_grouped = cutoffs_grouped[best_index_grouped]

print(f"Selected cutoff: {cutoff_grouped:.3f}")
print(f"Validation precision: {precision_grouped[best_index_grouped + 1]:.3f}")
print(f"Validation recall: {recall_grouped[best_index_grouped + 1]:.3f}")
print(f"Validation F1: {f1_grouped[best_index_grouped]:.3f}")
```

## Model 3
predicts whether a book is in the top 20% by units sold, using author rating, genre, and publisher. Average book rating and direct sales fields are excluded because they may reflect post-sales information.

```python
# Label the top 20% by units sold.
units_sold_cutoff_top20 = books["units sold"].quantile(0.80)
books["Best_Seller_Top20"] = (
    books["units sold"] >= units_sold_cutoff_top20
).astype(int)

# Exclude average rating because its timing is unclear.
predictors_top20 = ["Author_Rating", "genre", "Publisher"]

X_top20 = books[predictors_top20].copy()
y_top20 = books["Best_Seller_Top20"].copy()

print(f"Top-20% units-sold cutoff: {units_sold_cutoff_top20:,.0f}")
print("Class counts (0 = below cutoff, 1 = top 20%):")
display(y_top20.value_counts().sort_index())
```

The target has 214 positives. Split by author, keeping each author’s books together, so the test set measures performance on unseen authors.

### Split and encode the new model’s data

```python
from sklearn.model_selection import StratifiedGroupKFold
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer

# Use author only to keep each author's books together during splitting.
author_groups = books.loc[X_top20.index, "Author"].fillna("Unknown").astype(str)

# Reserve one group-stratified fold as the test set.
outer_split = StratifiedGroupKFold(
    n_splits=5, shuffle=True, random_state=42
)
train_validation_positions, test_positions = next(
    outer_split.split(X_top20, y_top20, groups=author_groups)
)

# Split the remaining authors into training and validation sets.
inner_split = StratifiedGroupKFold(
    n_splits=5, shuffle=True, random_state=43
)
train_positions_relative, validation_positions_relative = next(
    inner_split.split(
        X_top20.iloc[train_validation_positions],
        y_top20.iloc[train_validation_positions],
        groups=author_groups.iloc[train_validation_positions]
    )
)

train_positions = train_validation_positions[train_positions_relative]
validation_positions = train_validation_positions[validation_positions_relative]

# Create the three data sets.
X_train_raw_top20 = X_top20.iloc[train_positions]
X_validation_raw_top20 = X_top20.iloc[validation_positions]
X_test_raw_top20 = X_top20.iloc[test_positions]

y_train_top20 = y_top20.iloc[train_positions].to_numpy()
y_validation_top20 = y_top20.iloc[validation_positions].to_numpy()
y_test_top20 = y_top20.iloc[test_positions].to_numpy()

# One-hot encode the three categorical predictors, fitting on training data only.
preprocessor_top20 = ColumnTransformer([
    ("categorical", OneHotEncoder(handle_unknown="ignore"), predictors_top20)
])

X_train_top20 = preprocessor_top20.fit_transform(X_train_raw_top20)
X_validation_top20 = preprocessor_top20.transform(X_validation_raw_top20)
X_test_top20 = preprocessor_top20.transform(X_test_raw_top20)

print("Training rows/features:", X_train_top20.shape)
print("Validation rows/features:", X_validation_top20.shape)
print("Test rows/features:", X_test_top20.shape)
print("Positive shares — train, validation, test:",
      round(y_train_top20.mean(), 3),
      round(y_validation_top20.mean(), 3),
      round(y_test_top20.mean(), 3))
```

The author-grouped split keeps authors separate across sets. Positive share varies because each author’s books stay together. This is expected and makes the test more relevant to unseen authors.

Weighted loss fell from 1.0769 to 1.0425. Select the cutoff on validation data only, using the highest F1.

### Select the cutoff

```python
from sklearn.metrics import precision_recall_curve

# Predict probabilities for validation books.
validation_probabilities_top20 = sigmoid(
    X_validation_top20 @ weights_top20 + bias_top20
)

# Measure precision and recall across possible cutoffs.
precision_top20, recall_top20, cutoffs_top20 = precision_recall_curve(
    y_validation_top20, validation_probabilities_top20
)

# Calculate F1 for each cutoff.
f1_top20 = (
    2 * precision_top20[1:] * recall_top20[1:] /
    (precision_top20[1:] + recall_top20[1:] + 1e-12)
)

# Choose the cutoff with the highest validation F1.
best_index_top20 = np.argmax(f1_top20)
cutoff_top20 = cutoffs_top20[best_index_top20]

print(f"Selected cutoff: {cutoff_top20:.3f}")
print(f"Validation precision: {precision_top20[best_index_top20 + 1]:.3f}")
print(f"Validation recall: {recall_top20[best_index_top20 + 1]:.3f}")
print(f"Validation F1: {f1_top20[best_index_top20]:.3f}")
```

Compare the original logistic regression (Author_Rating, genre, Publisher) with a simpler one (Author_Rating, genre), using author-grouped validation. Assess with PR-AUC (primary), plus F1 and balanced accuracy.

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import StratifiedGroupKFold, cross_validate

# Use complete records for the target, predictors, and author grouping.
model_data = books[
    ["Author", "Author_Rating", "genre", "Publisher", "Best_Seller_Top20"]
].dropna().copy()

y_compare = model_data["Best_Seller_Top20"].astype(int)
author_groups = model_data["Author"]

# Compare the previous feature set with the reduced candidate.
feature_sets = {
    "Previous model — three features": [
        "Author_Rating", "genre", "Publisher"
    ],
    "Candidate model — two features": [
        "Author_Rating", "genre"
    ]
}

cv = StratifiedGroupKFold(n_splits=5, shuffle=True, random_state=42)

scoring = {
    "PR-AUC": "average_precision",
    "ROC-AUC": "roc_auc",
    "Balanced accuracy": "balanced_accuracy",
    "F1": "f1"
}

for model_name, predictors in feature_sets.items():
    X_compare = model_data[predictors]

    preprocess = ColumnTransformer([
        ("categorical", OneHotEncoder(handle_unknown="ignore"), predictors)
    ])

    model = Pipeline([
        ("preprocess", preprocess),
        ("logistic_regression", LogisticRegression(
            penalty="l2",
            C=1.0,
            class_weight=None,
            max_iter=2000
        ))
    ])

    results = cross_validate(
        model,
        X_compare,
        y_compare,
        groups=author_groups,
        cv=cv,
        scoring=scoring,
        error_score="raise"
    )

    print(f"\n{model_name}")
    for metric in scoring:
        fold_scores = results[f"test_{metric}"]
        print(f"{metric}: {fold_scores.mean():.3f} ± {fold_scores.std():.3f}")
```

PR-AUC is nearly identical (0.249 for three features, 0.248 for two), and the two-feature model’s slightly higher ROC-AUC is within fold-to-fold variation. Prefer the simpler model, but there is no clear performance gain. F1 is zero because the 0.50 cutoff predicted no top-20% books. PR-AUC shows some ranking ability, so next, pick a lower cutoff on training data within each fold and evaluate it on held-out authors.

```python
import numpy as np

from sklearn.base import clone
from sklearn.metrics import (
    average_precision_score,
    balanced_accuracy_score,
    f1_score,
    precision_score,
    recall_score,
    roc_auc_score,
    precision_recall_curve
)
from sklearn.model_selection import StratifiedGroupKFold, cross_val_predict
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression

# Compare the two feature sets with author-grouped outer folds.
feature_sets = {
    "Previous model — three features": [
        "Author_Rating", "genre", "Publisher"
    ],
    "Candidate model — two features": [
        "Author_Rating", "genre"
    ]
}

outer_cv = StratifiedGroupKFold(n_splits=5, shuffle=True, random_state=42)
results_by_model = {}

for model_name, predictors in feature_sets.items():
    X = model_data[predictors]
    fold_results = []

    for train_idx, test_idx in outer_cv.split(X, y_compare, author_groups):
        X_train, X_test = X.iloc[train_idx], X.iloc[test_idx]
        y_train, y_test = y_compare.iloc[train_idx], y_compare.iloc[test_idx]
        groups_train = author_groups.iloc[train_idx]

        # Build the logistic regression pipeline.
        preprocess = ColumnTransformer([
            ("categorical", OneHotEncoder(handle_unknown="ignore"), predictors)
        ])
        model = Pipeline([
            ("preprocess", preprocess),
            ("logistic_regression", LogisticRegression(
                penalty="l2",
                C=1.0,
                class_weight=None,
                max_iter=2000
            ))
        ])

        # Choose the cutoff from out-of-fold predictions within training data.
        inner_cv = StratifiedGroupKFold(
            n_splits=4, shuffle=True, random_state=42
        )
        train_oof_probabilities = cross_val_predict(
            clone(model),
            X_train,
            y_train,
            groups=groups_train,
            cv=inner_cv,
            method="predict_proba"
        )[:, 1]

        precision_values, recall_values, thresholds = precision_recall_curve(
            y_train, train_oof_probabilities
        )
        f1_values = (
            2 * precision_values[:-1] * recall_values[:-1]
            / (precision_values[:-1] + recall_values[:-1] + 1e-12)
        )
        selected_cutoff = thresholds[np.argmax(f1_values)]

        # Evaluate on authors held out from the outer training fold.
        model.fit(X_train, y_train)
        test_probabilities = model.predict_proba(X_test)[:, 1]
        test_predictions = (test_probabilities >= selected_cutoff).astype(int)

        fold_results.append({
            "cutoff": selected_cutoff,
            "PR-AUC": average_precision_score(y_test, test_probabilities),
            "ROC-AUC": roc_auc_score(y_test, test_probabilities),
            "Balanced accuracy": balanced_accuracy_score(y_test, test_predictions),
            "Precision": precision_score(y_test, test_predictions, zero_division=0),
            "Recall": recall_score(y_test, test_predictions, zero_division=0),
            "F1": f1_score(y_test, test_predictions, zero_division=0)
        })

    results_by_model[model_name] = fold_results

    print(f"\n{model_name}")
    for metric in fold_results[0]:
        values = [fold[metric] for fold in fold_results]
        print(f"{metric}: {np.mean(values):.3f} ± {np.std(values):.3f}")
```

The two-feature model (Author_Rating + genre) is the better generalized candidate: higher mean balanced accuracy, precision, and F1, with less recall variation. The three-feature model catches more positives but adds false positives and has lower F1. PR-AUC is essentially tied (0.248 vs. 0.249), and both are only modestly above the 0.20 baseline, so neither separates top sellers well. The 0.234 average cutoff shouldn’t be reused as final; select the final cutoff from training data. For the report: describe the two-feature model as the more balanced candidate, with modest performance and limited evidence of strong generalization.
